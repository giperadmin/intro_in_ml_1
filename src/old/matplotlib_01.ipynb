```python
import numpy as np

# Диапазон от -8.5 до 8.5
x = np.linspace(-8.5, 8.5, 2000)

# Полином с корнями в ±1, ±2, ±3, ±4, ±5, ±6, ±7, ±8
# y = (x^2 - 1)(x^2 - 4)(x^2 - 9)...(x^2 - 64)
y = (x**2 - 1) * (x**2 - 4) * (x**2 - 9) * (x**2 - 16) * \
    (x**2 - 25) * (x**2 - 36) * (x**2 - 49) * (x**2 - 64)

plt.figure(figsize=(10, 6))
plt.plot(x, y, color='crimson', linewidth=1.5)

# Ограничим ось Y, чтобы видеть красивые колебания в центре.
# Без этого края графика улетят в миллионы и сожмут центр в плоскую линию.
plt.ylim(-1e6, 1e6)

plt.grid(True, linestyle='--', alpha=0.7)
plt.axhline(0, color='black', linewidth=1)
plt.title(r"Полином 16-й степени: $y = \prod_{k=1}^{8} (x^2 - k^2)$", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.show()
```
