# ML3. Часть 6. Feature Selection / Отбор признаков

В этом ноутбуке решается 6-я часть задания из README. Целевая переменная проекта - `price`, то есть решается задача регрессии.

Что делаем:

1. Обучаем `Lasso`-регрессию на нормализованных признаках. Разбиваем данные по `created` в пропорции 60/20/20: `train/validation/test`.
2. Сортируем признаки по весовым коэффициентам `Lasso`, берем top-10 и сравниваем качество.
3. Реализуем простой отбор по nan-ratio и correlation, берем top-10, переобучаем модель и измеряем качество.
4. Реализуем `permutation importance`, берем top-10, переобучаем модель и измеряем качество.
5. Пробуем импортировать `shap`, выбираем top-10 признаков по SHAP и переобучаем модель.
6. Сравниваем методы по скорости, метрикам и стабильности.

## 1. Импорт библиотек и загрузка данных

Используем `pandas`, `numpy` и инструменты `sklearn`. `shap` импортируется позже в отдельном блоке: в текущем окружении библиотека может быть не установлена, поэтому обработаем этот случай аккуратно.

```python
from pathlib import Path
from time import perf_counter
import warnings

import numpy as np
import pandas as pd
from sklearn.exceptions import ConvergenceWarning
from sklearn.inspection import permutation_importance
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore", category=ConvergenceWarning)

RANDOM_STATE = 21
TARGET = "price"
DATE_FIELD = "created"
DATA_PATH = Path(r"C:\21school\ML3_Validation_ID_1254800-1\src\data\train_df_processed_0.75.csv")

df = pd.read_csv(DATA_PATH)
df[DATE_FIELD] = pd.to_datetime(df[DATE_FIELD])

print(f"Размер датасета: {df.shape}")
display(df.head())
display(df.dtypes)
```

## 2. Разбиение по `created` в пропорции 60/20/20

По заданию нужно использовать метод разбиения по полю `created` с пропорцией 60/20/20: `train/validation/test`.

Логика:

* сортируем данные по времени;
* самые ранние 60% отправляем в `train`;
* следующие 20% - в `validation`;
* последние 20% - в `test`.

Такой вариант имитирует production-сценарий: модель обучается на прошлом и проверяется на более новых данных.

```python
def split_by_date_ratio(
    data: pd.DataFrame,
    date_field: str,
    train_size: float = 0.60,
    validation_size: float = 0.20,
    test_size: float = 0.20,
) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Делит DataFrame по дате на train/validation/test в заданных долях."""
    if not np.isclose(train_size + validation_size + test_size, 1.0):
        raise ValueError("Сумма train_size, validation_size и test_size должна быть равна 1")

    prepared = data.copy()
    prepared[date_field] = pd.to_datetime(prepared[date_field])
    prepared = prepared.sort_values(date_field).reset_index(drop=True)

    n_rows = len(prepared)
    train_end = int(n_rows * train_size)
    validation_end = int(n_rows * (train_size + validation_size))

    train = prepared.iloc[:train_end].copy()
    validation = prepared.iloc[train_end:validation_end].copy()
    test = prepared.iloc[validation_end:].copy()

    return train, validation, test


df_train, df_validation, df_test = split_by_date_ratio(df, DATE_FIELD)

print("train:", df_train.shape, df_train[DATE_FIELD].min(), "-", df_train[DATE_FIELD].max())
print("validation:", df_validation.shape, df_validation[DATE_FIELD].min(), "-", df_validation[DATE_FIELD].max())
print("test:", df_test.shape, df_test[DATE_FIELD].min(), "-", df_test[DATE_FIELD].max())
```

## 3. Подготовка признаков

Цель - `price`. Из признаков исключаем:

* `price` - target;
* `created` - поле для временного разбиения;
* `interest` - отдельная метка интереса из исходной задачи, не базовый признак квартиры для предсказания цены.

Оставшиеся числовые признаки используются для обучения.

```python
excluded_columns = {TARGET, DATE_FIELD, "interest"}
feature_columns = [
    column
    for column in df_train.select_dtypes(include=np.number).columns
    if column not in excluded_columns
]

X_train = df_train[feature_columns]
y_train = df_train[TARGET]

X_validation = df_validation[feature_columns]
y_validation = df_validation[TARGET]

X_test = df_test[feature_columns]
y_test = df_test[TARGET]

print(f"Количество признаков: {len(feature_columns)}")
print(feature_columns)
display(X_train.describe().T.head(10))
```

## 4. Общие функции для обучения и оценки качества

Модель: `Lasso` с нормализацией признаков через `StandardScaler`.

Метрики:

* `RMSE` - корень из средней квадратичной ошибки;
* `MAE` - средняя абсолютная ошибка;
* `R2` - коэффициент детерминации.

Основной ориентир - `RMSE` и `MAE`: они измеряются в тех же единицах, что и `price`.

```python
def make_lasso_model(alpha: float = 1.0) -> Pipeline:
    """Создает pipeline: StandardScaler + Lasso."""
    return Pipeline(
        steps=[
            ("scaler", StandardScaler()),
            (
                "model",
                Lasso(
                    alpha=alpha,
                    random_state=RANDOM_STATE,
                    max_iter=50_000,
                ),
            ),
        ]
    )


def regression_metrics(y_true, y_pred) -> dict:
    """Считает метрики регрессии."""
    return {
        "rmse": float(np.sqrt(mean_squared_error(y_true, y_pred))),
        "mae": float(mean_absolute_error(y_true, y_pred)),
        "r2": float(r2_score(y_true, y_pred)),
    }


def fit_and_evaluate(feature_subset: list[str], method_name: str, alpha: float = 1.0) -> tuple[Pipeline, dict]:
    """Обучает Lasso на выбранных признаках и возвращает модель с метриками."""
    start_time = perf_counter()
    model = make_lasso_model(alpha=alpha)
    model.fit(X_train[feature_subset], y_train)
    elapsed = perf_counter() - start_time

    validation_pred = model.predict(X_validation[feature_subset])
    test_pred = model.predict(X_test[feature_subset])

    metrics = {
        "method": method_name,
        "n_features": len(feature_subset),
        "features": feature_subset,
        "fit_time_sec": elapsed,
        **{f"validation_{k}": v for k, v in regression_metrics(y_validation, validation_pred).items()},
        **{f"test_{k}": v for k, v in regression_metrics(y_test, test_pred).items()},
    }

    return model, metrics


results = []
```

## 5. Baseline: Lasso на всех нормализованных признаках

Сначала обучим `Lasso` на всех доступных признаках. Это базовая модель, с которой будем сравнивать варианты после отбора признаков.

```python
lasso_all_model, lasso_all_metrics = fit_and_evaluate(
    feature_columns,
    method_name="Lasso: все признаки",
    alpha=1.0,
)
results.append(lasso_all_metrics)

display(pd.DataFrame([lasso_all_metrics]).drop(columns="features"))
```

## 6. Top-10 по коэффициентам Lasso

Так как признаки нормализуются через `StandardScaler`, абсолютные значения коэффициентов `Lasso` можно использовать как оценку важности признаков.

Берем top-10 признаков по `abs(coef)` и переобучаем модель только на них.

```python
lasso_coefficients = pd.Series(
    lasso_all_model.named_steps["model"].coef_,
    index=feature_columns,
    name="coefficient",
)
lasso_importance = (
    lasso_coefficients
    .abs()
    .sort_values(ascending=False)
    .rename("abs_coefficient")
    .to_frame()
)
lasso_top10_features = lasso_importance.head(10).index.tolist()

lasso_top10_model, lasso_top10_metrics = fit_and_evaluate(
    lasso_top10_features,
    method_name="Top-10 по коэффициентам Lasso",
    alpha=1.0,
)
results.append(lasso_top10_metrics)

print("Top-10 признаков по коэффициентам Lasso:")
display(lasso_importance.head(10))
display(pd.DataFrame([lasso_top10_metrics]).drop(columns="features"))
```

## 7. Простой отбор по nan-ratio и correlation

Реализуем простой filter-подход:

1. считаем долю пропусков (`nan_ratio`) в каждом признаке;
2. считаем абсолютную корреляцию признака с target `price` на train;
3. исключаем признаки с большим `nan_ratio`;
4. сортируем оставшиеся признаки по `abs(correlation)`;
5. берем top-10.

```python
def select_by_nan_ratio_and_correlation(
    X: pd.DataFrame,
    y: pd.Series,
    top_n: int = 10,
    max_nan_ratio: float = 0.20,
) -> tuple[list[str], pd.DataFrame]:
    """Выбирает top-N признаков по nan-ratio и корреляции с target."""
    rows = []

    for column in X.columns:
        nan_ratio = X[column].isna().mean()
        correlation = X[column].corr(y)

        rows.append(
            {
                "feature": column,
                "nan_ratio": nan_ratio,
                "correlation": correlation,
                "abs_correlation": abs(correlation) if pd.notna(correlation) else 0.0,
            }
        )

    scores = pd.DataFrame(rows).set_index("feature")
    scores = scores[scores["nan_ratio"] <= max_nan_ratio]
    scores = scores.sort_values("abs_correlation", ascending=False)

    return scores.head(top_n).index.tolist(), scores


filter_top10_features, filter_scores = select_by_nan_ratio_and_correlation(
    X_train,
    y_train,
    top_n=10,
    max_nan_ratio=0.20,
)

filter_top10_model, filter_top10_metrics = fit_and_evaluate(
    filter_top10_features,
    method_name="Top-10 по nan-ratio и correlation",
    alpha=1.0,
)
results.append(filter_top10_metrics)

print("Top-10 признаков по nan-ratio и correlation:")
display(filter_scores.head(10))
display(pd.DataFrame([filter_top10_metrics]).drop(columns="features"))
```

## 8. Permutation importance

`permutation importance` измеряет, насколько ухудшается качество модели, если случайно перемешать значения одного признака. Если после перемешивания качество сильно падает, значит признак важен.

Считаем importance на validation-части для baseline-модели `Lasso: все признаки`, затем берем top-10 и переобучаем модель.

```python
start_time = perf_counter()
permutation_result = permutation_importance(
    lasso_all_model,
    X_validation[feature_columns],
    y_validation,
    scoring="neg_root_mean_squared_error",
    n_repeats=20,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)
permutation_time = perf_counter() - start_time

permutation_scores = pd.DataFrame(
    {
        "importance_mean": permutation_result.importances_mean,
        "importance_std": permutation_result.importances_std,
    },
    index=feature_columns,
).sort_values("importance_mean", ascending=False)

permutation_top10_features = permutation_scores.head(10).index.tolist()

permutation_top10_model, permutation_top10_metrics = fit_and_evaluate(
    permutation_top10_features,
    method_name="Top-10 по permutation importance",
    alpha=1.0,
)
permutation_top10_metrics["selection_time_sec"] = permutation_time
results.append(permutation_top10_metrics)

print("Top-10 признаков по permutation importance:")
display(permutation_scores.head(10))
display(pd.DataFrame([permutation_top10_metrics]).drop(columns="features"))
```

## 9. SHAP

По заданию нужно импортировать `shap` и переобучить модель на top-10 признаках по SHAP.

Если `shap` установлен, используем `shap.LinearExplainer` для baseline Lasso-модели. Если библиотека не установлена, ноутбук не падает: используется fallback для линейной модели - средний абсолютный вклад нормализованного признака `abs(X_scaled * coef)`. Для линейной модели это близко по смыслу к оценке вклада признаков.

```python
start_time = perf_counter()

try:
    import shap

    SHAP_AVAILABLE = True
except ImportError:
    shap = None
    SHAP_AVAILABLE = False

scaler = lasso_all_model.named_steps["scaler"]
lasso_model = lasso_all_model.named_steps["model"]
X_train_scaled = scaler.transform(X_train[feature_columns])
X_validation_scaled = scaler.transform(X_validation[feature_columns])

if SHAP_AVAILABLE:
    explainer = shap.LinearExplainer(lasso_model, X_train_scaled)
    shap_values = explainer.shap_values(X_validation_scaled)
    shap_importance_values = np.abs(shap_values).mean(axis=0)
    shap_source = "shap.LinearExplainer"
else:
    shap_importance_values = np.abs(X_validation_scaled * lasso_model.coef_).mean(axis=0)
    shap_source = "fallback: abs(X_scaled * coef)"

shap_time = perf_counter() - start_time

shap_scores = pd.DataFrame(
    {"mean_abs_shap": shap_importance_values},
    index=feature_columns,
).sort_values("mean_abs_shap", ascending=False)

shap_top10_features = shap_scores.head(10).index.tolist()

shap_top10_model, shap_top10_metrics = fit_and_evaluate(
    shap_top10_features,
    method_name="Top-10 по SHAP",
    alpha=1.0,
)
shap_top10_metrics["selection_time_sec"] = shap_time
results.append(shap_top10_metrics)

print("SHAP доступен:", SHAP_AVAILABLE)
print("Источник важности:", shap_source)
print("Top-10 признаков по SHAP/fallback:")
display(shap_scores.head(10))
display(pd.DataFrame([shap_top10_metrics]).drop(columns="features"))
```

## 10. Сравнение методов: метрики, скорость, стабильность

Стабильность здесь оценивается как пересечение top-10 признаков метода с top-10 признаками baseline Lasso по коэффициентам. Это простая практичная оценка: чем больше пересечение, тем ближе метод к встроенному отбору `Lasso`.

Также сравниваем скорость отбора/обучения и метрики на validation/test.

```python
baseline_top_set = set(lasso_top10_features)

def overlap_with_lasso_top10(features: list[str]) -> float:
    """Доля пересечения с top-10 признаками по коэффициентам Lasso."""
    return len(set(features) & baseline_top_set) / len(baseline_top_set)


summary = pd.DataFrame(results)
summary["selection_time_sec"] = summary.get("selection_time_sec", np.nan)
summary["total_time_sec"] = summary["fit_time_sec"] + summary["selection_time_sec"].fillna(0)
summary["stability_vs_lasso_top10"] = summary["features"].apply(overlap_with_lasso_top10)

metrics_columns = [
    "method",
    "n_features",
    "fit_time_sec",
    "selection_time_sec",
    "total_time_sec",
    "validation_rmse",
    "validation_mae",
    "validation_r2",
    "test_rmse",
    "test_mae",
    "test_r2",
    "stability_vs_lasso_top10",
]

display(summary[metrics_columns].sort_values("validation_rmse"))

print("Выбранные признаки по методам:")
for _, row in summary.iterrows():
    print(f"\n{row['method']}:")
    print(row["features"])
```

## 11. Итоговые выводы

* `Lasso` на всех признаках - базовая модель. Она использует всю доступную информацию, но может быть менее интерпретируемой.
* Top-10 по коэффициентам `Lasso` - встроенный (`embedded`) отбор. Он быстрый и естественно связан с используемой моделью.
* Отбор по nan-ratio и correlation - простой `filter`-метод. Он очень быстрый, но видит только индивидуальную линейную связь признака с `price` и не учитывает взаимодействия признаков.
* `permutation importance` оценивает влияние признака на реальную метрику модели. Метод более дорогой, зато ближе к практическому влиянию признака на качество.
* `SHAP` дает интерпретируемую оценку вкладов признаков. Если библиотека `shap` недоступна, в ноутбуке используется линейный fallback, чтобы сохранить исполняемость решения.

Лучший метод стоит выбирать по validation-метрике, а test использовать как финальную проверку. Если несколько методов дают близкое качество, предпочтительнее более простой и стабильный набор признаков.
