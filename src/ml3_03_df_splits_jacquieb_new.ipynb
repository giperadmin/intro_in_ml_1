## Импорт библиотек, методов и датафрейма.

```python
import pandas as pd

from ml3_03_data_split_methods_jacquieb import (
    split_random_train_test,
    split_random_train_validation_test,
    split_by_date_train_test,
    split_by_dates_train_validation_test,
    split_by_date_ratio_train_test,
    split_by_date_ratio_train_validation_test,
    )

from ml3_03_split_visualization_jacquieb import (
    visualize_train_test_split,
    visualize_train_valid_test_split,
    )

RANDOM_STATE = 21
DATA_PATH = (r".\data\train_df_processed_full.csv")

df = pd.read_csv(DATA_PATH)
df["created"] = pd.to_datetime(df["created"])

df['Elevator'].sum()

# df_train, df_test = split_random_train_test(df, test_size=0.2)
```

## Разделить данные случайно на 2 части с параметром test_size (доля от 0 до 1), вернуть обучающую и тестовую выборки.

```python
df_train, df_test = split_random_train_test(df, test_size=0.2)
visualize_train_test_split(df_train[['price', 'created']], df_test[['price', 'created']])
```

## Случайно разделить данные на 3 части с параметрами validation_size и test_size, вернуть train, validation и test выборки.

```python
df_train, df_valid, df_test = split_random_train_validation_test(df, test_size=0.2, validation_size=0.2)
visualize_train_valid_test_split(df_train, df_valid, df_test)
```

## Разделить данные на 2 части с параметром date_split, вернуть train и test выборки, разделенные по параметру date_split.

```python
df_train, df_test = split_by_date_train_test(df, date_split='2016-05-15')
visualize_train_test_split(df_train[['price', 'created']], df_test[['price', 'created']])
```

## Разделить данные на 3 части с параметрами validation_date и test_date, вернуть train, validation и test выборки, разделенные по входным параметрам.

```python
df_train, df_valid, df_test = split_by_dates_train_validation_test(df, validation_date='2016-04-28', test_date='2016-06-06')
visualize_train_valid_test_split(df_train, df_valid, df_test)
```

## Bonus. Разделить на две части по дате с параметром test_size (доля от 0 до 1).

```python
df_train, df_test = split_by_date_ratio_train_test(df,test_size=0.38)
visualize_train_test_split(df_train, df_test)
```

## Bonus. Разделить на три части по дате с параметрами validation_size и test_size (доля от 0 до 1).

```python
df_train, df_valid, df_test = split_by_date_ratio_train_validation_test(df, validation_size=0.38, test_size=0.38)
visualize_train_valid_test_split(df_train, df_valid, df_test)
```
